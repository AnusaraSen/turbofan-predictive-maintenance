```python
import sys
sys.path.insert(0, "../src")

import matplotlib.pyplot as plt
from data_loader import load_train_data
from preprocessing import compute_rul, fit_regime_clusters, assign_regimes

train_df = load_train_data("FD002")   # explicit, not the default
train_df = compute_rul(train_df)
kmeans = fit_regime_clusters(train_df, n_regimes=6)
train_df = assign_regimes(train_df, kmeans)

plt.scatter(train_df["op_setting_1"], train_df["op_setting_2"], c=train_df["regime_id"], cmap="tab10", s=2)
plt.xlabel("op_setting_1")
plt.ylabel("op_setting_2")
plt.colorbar(label="regime_id")
plt.show()
```
